# ***Question 2***

We have created the set P1 and defined the error terms for all prices in the set.
Here we show the errors around zero. We also plot them in the figure below.

```python
p1_range = model.P1set() 
errors = model.Q2errors(p1_range)
print("The errors for p1 in P1 around zero:")
print(errors[16:18])
```

```python
fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot(1,1,1)
ax.set_title('Errors in the market clearing price ')
ax.plot(p1_range,errors,'-s',label='Market clearing error')

ax.set_xlabel('$p_1$')
ax.set_ylabel('$\epsilon$')
ax.set_xlim([0.5, 2.5])
ax.legend(frameon=True)
```
