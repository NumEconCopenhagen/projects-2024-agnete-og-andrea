# Model description

We choose an OLG model with the following attributes: 

- **Time** is discrete and infinite
- **Population** is constant. Implying a population growth rate, $n=0$.
- **Households** live for two periods: young and old. As young a household supplies labor exogenously, $N_{t}=1$, and earns an after tax wage $(1-\tau_w)w_{t}N_{t}$.<br>
Consumption as young and old are denoted by $C_{1t}$ and $C_{2t+1}$. <br>
The young in period t pay contributions to the old in period t, denoted by $d_{t}$. Agents do not work when old, and live off their gross savings and contributions from the young, denoted by $b_{t}$  <br>
The after-tax return on saving is $(1-\tau_{r})r_{t+1}$.<br>
The households maximize the following problem:<br>

$$
\begin{aligned}
U & =\max_{s_{t}\in[0,1]}\frac{C_{1t}^{1-\sigma}}{1-\sigma}+\beta\frac{C_{2t+1}^{1-\sigma}}{1-\sigma},\,\,\,\beta > -1, \sigma > 0\\
 & \text{s.t.}\\
 & S_{t}=s_{t}[(1-\tau_{w})w_{t}N_{t}-d_t], s_t\in[0,1]\\
 & C_{1t}=(1-s_{t})[(1-\tau_{w})w_{t}N_{t}-d_t]\\
 & C_{2t+1}=(1+(1-\tau_{r})r_{t+1})S_{t}+b_{t+1}\\
 & b_t = (1+n)d_t\\
\end{aligned}
$$ 

**Firms:** Firms rent capital $K_{t-1}$ at the rental rate $r_{t}^{K}$,
and hires labor $L_{t}$ at the wage rate $w_{t}$. Firms have access
to the production function
    
$$
\begin{aligned}
Y_{t}=F(K_{t-1},L_{t})=(\alpha K_{t-1}^{-\theta}+(1-\alpha)L_{t}^{-\theta})^{\frac{1}{-\theta}},\,\,\,\theta>-1,\alpha\in(0,1)
\end{aligned}
$$
    
Profits are
    
$$
\begin{aligned}
\Pi_{t}=Y_{t}-w_{t}L_{t}-r_{t}^{K}K_{t-1}
\end{aligned}
$$

**Government:** Choose public consumption, $G_{t}$, and tax rates $\tau_w \in [0,1]$ and $\tau_r \in [0,1]$. Total tax revenue is
    
$$
\begin{aligned}
T_{t} &=\tau_r r_{t} (K_{t-1}+B_{t-1})+\tau_w w_{t}N_{t}+d_t   
\end{aligned}
$$
    
Government debt accumulates according to
    
$$
\begin{aligned}
B_{t} &=(1+r^b_{t})B_{t-1}-T_{t}+G_{t}+b_t 
\end{aligned}
$$
       
A *balanced budget* implies $G_{t}=T_{t}-r_{t}B_{t-1}$.

**Capital:** Depreciates with a rate of $\delta \in [0,1]$.
